## Advanced Options

Beyond local editing two things are quite important: editing the notebooks in the markdown format and running Jupyter remotely.
The latter matters when we want to run the code on a faster server.
The former matters since Jupyter's native ipynb format stores a lot of auxiliary data that is
irrelevant to the content,
mostly related to how and where the code is run.
This is confusing for Git, making
reviewing contributions very difficult.
Fortunately there is an alternative---native editing in the markdown format.

### Markdown Files in Jupyter

If you wish to contribute to the content of this book, you need to modify the
source file (md file, not ipynb file) on GitHub.
Using the notedown plugin we
can modify notebooks in the md format directly in Jupyter.


First, install the notedown plugin, run the Jupyter Notebook, and load the plugin:

```
pip install d2l-notedown  # You may need to uninstall the original notedown.
jupyter notebook --NotebookApp.contents_manager_class='notedown.NotedownContentsManager'
```

You may also turn on the notedown plugin by default whenever you run the Jupyter Notebook.
First, generate a Jupyter Notebook configuration file (if it has already been generated, you can skip this step).

```
jupyter notebook --generate-config
```

Then, add the following line to the end of the Jupyter Notebook configuration file (for Linux or macOS, usually in the path `~/.jupyter/jupyter_notebook_config.py`):

```
c.NotebookApp.contents_manager_class = 'notedown.NotedownContentsManager'
```

After that, you only need to run the `jupyter notebook` command to turn on the notedown plugin by default.

### Running Jupyter Notebooks on a Remote Server

Sometimes, you may want to run Jupyter notebooks on a remote server and access it through a browser on your local computer. If Linux or macOS is installed on your local machine (Windows can also support this function through third-party software such as PuTTY), you can use port forwarding:

```
ssh myserver -L 8888:localhost:8888
```

The above string `myserver` is the address of the remote server.
Then we can use http://localhost:8888 to access the remote server `myserver` that runs Jupyter notebooks. We will detail on how to run Jupyter notebooks on AWS instances
later in this appendix.

### Timing

We can use the `ExecuteTime` plugin to time the execution of each code cell in Jupyter notebooks.
Use the following commands to install the plugin:

```
pip install jupyter_contrib_nbextensions
jupyter contrib nbextension install --user
jupyter nbextension enable execute_time/ExecuteTime
```

````python
import os, pandas as pd

data_dir = "/content/drive/MyDrive/task1/training"
train_path = "/content/drive/MyDrive/task1/train_data.csv"

train_df = pd.read_csv(train_path)
train_df["id"] = train_df["id"].astype(str).str.upper().str.strip()

records = []

for root, _, filenames in os.walk(data_dir):
    for f in filenames:
        if f.endswith('.wav'):
            # Use the real full path directly (preserves actual casing)
            path = os.path.join(root, f)

            speaker_id = f.split("_")[0].upper()
            row = train_df[train_df["id"] == speaker_id]
            if not row.empty:
                label = int(row["class"].values[0]) - 1
                records.append({
                    "filename": f,
                    "filepath": path,   # ✅ uses real path casing
                    "label": label
                })

labels_df = pd.DataFrame(records)
labels_df.to_csv("labels.csv", index=False)
print("✅ labels.csv created successfully with", len(labels_df), "rows.")
labels_df.head()
````

````python
import torchaudio, torch
from torch.utils.data import Dataset, DataLoader, random_split
from transformers import Wav2Vec2Processor

class AudioDataset(Dataset):
    def __init__(self, csv_path, processor_name="facebook/wav2vec2-base"):
        self.df = pd.read_csv(csv_path)
        # self.audio_dir is no longer needed as 'filepath' column is used directly
        self.processor = Wav2Vec2Processor.from_pretrained(processor_name)

    def __len__(self): return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        path = row['filepath']  # Use the correct full path from the DataFrame
        waveform, sr = torchaudio.load(path)
        waveform = waveform.mean(dim=0)  # mono
        if sr != 16000:
            waveform = torchaudio.functional.resample(waveform, sr, 16000)
        waveform = waveform / waveform.abs().max()  # normalize
        # Wrap waveform in a list and specify max_length to ensure attention_mask is generated
        inputs = self.processor([waveform], sampling_rate=16000, return_tensors="pt",
                                padding="max_length", max_length=80000, truncation=True)
        return {
            "input_values": inputs.input_values.squeeze(0),
            "attention_mask": inputs.attention_mask.squeeze(0),
            "label": torch.tensor(row['label'])
        }

# Instantiate the dataset, removing the redundant data_dir parameter
dataset = AudioDataset("labels.csv")
len(dataset)
````

````python
from torch.utils.data import DataLoader, random_split, Subset

# SUPER SIMPLE - just take first 50 samples
small_dataset = Subset(dataset, range(50))

# Then do normal train/val split on the small dataset
train_size = int(0.8 * len(small_dataset))
val_size = len(small_dataset) - train_size
train_ds, val_ds = random_split(small_dataset, [train_size, val_size])

train_loader = DataLoader(train_ds, batch_size=4, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=4)

print(f"Using {len(small_dataset)} samples for quick testing")
````

````python
import torch
from transformers import Wav2Vec2ForSequenceClassification

num_labels = 5  # 5 classes total
model = Wav2Vec2ForSequenceClassification.from_pretrained(
    "facebook/wav2vec2-base",
    num_labels=num_labels,
    ignore_mismatched_sizes=True
).to("cuda")

optimizer = torch.optim.AdamW(model.parameters(), lr=1e-5)
````

````python
import torch
import gc
from tqdm import tqdm
import warnings
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

# Suppress torchaudio deprecation warnings
warnings.filterwarnings("ignore", category=UserWarning, module="torchaudio")
warnings.filterwarnings("ignore", category=UserWarning, module="torchio")

def train_with_memory_management(model, train_loader, val_loader, optimizer, epochs=5):
    # Initialize lists to store metrics
    train_losses = []
    val_losses = []
    val_accuracies = []
    val_f1_scores = []

    for epoch in range(epochs):
        # === TRAINING ===
        model.train()
        train_loss = 0

        # Clear memory before epoch
        torch.cuda.empty_cache()
        gc.collect()

        for batch_idx, batch in enumerate(tqdm(train_loader, desc=f"Epoch {epoch+1}")):
            try:
                input_values = batch["input_values"].to("cuda")
                attention_mask = batch["attention_mask"].to("cuda")
                labels = batch["label"].to("cuda")

                outputs = model(input_values=input_values,
                              attention_mask=attention_mask,
                              labels=labels)
                loss = outputs.loss

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                train_loss += loss.item()

                # Clear memory every N batches
                if batch_idx % 50 == 0:
                    torch.cuda.empty_cache()

            except RuntimeError as e:
                if "out of memory" in str(e):
                    print("Out of memory, skipping batch")
                    torch.cuda.empty_cache()
                    continue
                else:
                    raise e

        avg_train_loss = train_loss / len(train_loader)
        train_losses.append(avg_train_loss)
        print(f"Average train loss: {avg_train_loss:.4f}")

        # === VALIDATION ===
        model.eval()
        val_loss = 0
        all_predictions = []
        all_labels = []

        with torch.no_grad():
            for batch in tqdm(val_loader, desc="Validating"):
                try:
                    input_values = batch["input_values"].to("cuda")
                    attention_mask = batch["attention_mask"].to("cuda")
                    labels = batch["label"].to("cuda")

                    outputs = model(input_values=input_values,
                                  attention_mask=attention_mask,
                                  labels=labels)

                    loss = outputs.loss
                    val_loss += loss.item()

                    # Get predictions
                    logits = outputs.logits
                    predictions = torch.argmax(logits, dim=-1)

                    all_predictions.extend(predictions.cpu().numpy())
                    all_labels.extend(labels.cpu().numpy())

                except Exception as e:
                    print(f"Validation error: {e}")
                    continue

        avg_val_loss = val_loss / len(val_loader)
        val_accuracy = accuracy_score(all_labels, all_predictions)
        val_f1 = f1_score(all_labels, all_predictions, average='weighted')

        val_losses.append(avg_val_loss)
        val_accuracies.append(val_accuracy)
        val_f1_scores.append(val_f1)

        print(f"Validation Loss: {avg_val_loss:.4f}, Accuracy: {val_accuracy:.4f}, F1: {val_f1:.4f}")

        # Clear memory after epoch
        torch.cuda.empty_cache()
        gc.collect()

    # Return all collected metrics
    return train_losses, val_losses, val_accuracies, val_f1_scores
````

````python
import torch
from sklearn.metrics import f1_score
from tqdm import tqdm

def evaluate_first_n_batches(model, data_loader, n_batches=5):
    """
    Evaluate only the first n_batches from data_loader to collect metrics for plotting.
    Returns lists of per-batch metrics.
    """
    model.eval()
    batch_losses = []
    batch_accuracies = []
    batch_f1s = []

    with torch.no_grad():
        for i, batch in enumerate(tqdm(data_loader, desc="Evaluating first N batches")):
            if i >= n_batches:
                break

            input_values = batch["input_values"].to("cuda")
            attention_mask = batch["attention_mask"].to("cuda")
            labels = batch["label"].to("cuda")

            outputs = model(input_values=input_values,
                            attention_mask=attention_mask,
                            labels=labels)

            # Batch metrics
            batch_size = labels.size(0)
            batch_loss = outputs.loss.item()
            predictions = torch.argmax(outputs.logits, dim=-1)
            batch_acc = (predictions == labels).sum().item() / batch_size
            batch_f1 = f1_score(labels.cpu().numpy(), predictions.cpu().numpy(), average='weighted')

            batch_losses.append(batch_loss)
            batch_accuracies.append(batch_acc)
            batch_f1s.append(batch_f1)

            # Free memory
            del input_values, attention_mask, labels, outputs, predictions
            torch.cuda.empty_cache()

    return batch_losses, batch_accuracies, batch_f1s
````

````python
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
from google.colab import files

def plot_training_metrics_colab(train_loss, val_loss, train_acc, val_acc, val_f1):
    """
    Generates a single-page PDF with loss, accuracy, and F1 score plots, and downloads it in Colab.
    """
    save_path = "/content/training_summary.pdf"
    epochs = range(1, len(train_loss) + 1)

    with PdfPages(save_path) as pdf:
        fig, axs = plt.subplots(3, 1, figsize=(10, 12))

        # Loss
        axs[0].plot(epochs, train_loss, label='Train Loss', marker='o')
        axs[0].plot(epochs, val_loss, label='Validation Loss', marker='o')
        axs[0].set_title("Loss per Epoch")
        axs[0].set_xlabel("Epoch")
        axs[0].set_ylabel("Loss")
        axs[0].grid(True)
        axs[0].legend()

        # Accuracy
        axs[1].plot(epochs, train_acc, label='Train Accuracy', marker='o')
        axs[1].plot(epochs, val_acc, label='Validation Accuracy', marker='o')
        axs[1].set_title("Accuracy per Epoch")
        axs[1].set_xlabel("Epoch")
        axs[1].set_ylabel("Accuracy")
        axs[1].grid(True)
        axs[1].legend()

        # F1 Score
        axs[2].plot(epochs, val_f1, label='Validation F1', marker='o', color='green')
        axs[2].set_title("Validation F1 Score per Epoch")
        axs[2].set_xlabel("Epoch")
        axs[2].set_ylabel("F1 Score")
        axs[2].grid(True)
        axs[2].legend()

        plt.tight_layout()
        pdf.savefig(fig)
        plt.close()

    print(f"✅ PDF saved at {save_path}", flush=True)

    # Trigger download in Colab
    files.download(save_path)
````
